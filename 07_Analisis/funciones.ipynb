### Comparación de particiones

`comparar`: NMI y ARI entre dos etiquetados de proveedores, solo en los proveedores que tienen ambos (por clave)

```python
def comparar(a, b):
    """a, b: Series indexadas por clave de proveedor."""
    comun = a.dropna().index.intersection(b.dropna().index)
    x, y = a.loc[comun].astype(str), b.loc[comun].astype(str)
    return {"n": len(comun), "nmi": normalized_mutual_info_score(x, y), "ari": adjusted_rand_score(x, y)}
```
